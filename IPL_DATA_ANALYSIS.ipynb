```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (10, 6)

df = pd.read_csv("Match.csv(1).csv")

print("=" * 60)
print("IPL DATA ANALYSIS PROJECT")
print("=" * 60)

print("\nDataset Shape")
print(df.shape)

print("\nColumns")
print(df.columns.tolist())

print("\nFirst 5 Records")
print(df.head())

print("\nDataset Information")
print(df.info())

print("\nMissing Values")
print(df.isnull().sum())

df["winner"] = df["winner"].fillna("No Result")
df["player_of_match"] = df["player_of_match"].fillna("Not Awarded")

print("\nTop 10 Winning Teams")
top_winners = df["winner"].value_counts().head(10)
print(top_winners)

plt.figure(figsize=(10,5))
top_winners.plot(kind="bar")
plt.title("Top 10 Winning Teams")
plt.xlabel("Team")
plt.ylabel("Matches Won")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nTop 10 Player of the Match")
top_players = df["player_of_match"].value_counts().head(10)
print(top_players)

plt.figure(figsize=(10,5))
top_players.plot(kind="bar")
plt.title("Top 10 Player of the Match")
plt.xlabel("Player")
plt.ylabel("Awards")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nTop Toss Winners")
top_toss = df["toss_winner"].value_counts().head(10)
print(top_toss)

plt.figure(figsize=(10,5))
top_toss.plot(kind="bar")
plt.title("Top Toss Winning Teams")
plt.xlabel("Team")
plt.ylabel("Toss Wins")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nToss Decision")
decision = df["toss_decision"].value_counts()
print(decision)

plt.figure(figsize=(6,6))
plt.pie(decision,
        labels=decision.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Toss Decision Distribution")
plt.show()

print("\nTop 10 Venues")
venues = df["venue"].value_counts().head(10)
print(venues)

plt.figure(figsize=(12,6))
venues.plot(kind="bar")
plt.title("Top 10 IPL Venues")
plt.xlabel("Venue")
plt.ylabel("Matches")
plt.xticks(rotation=80)
plt.tight_layout()
plt.show()

print("\nTop 10 Cities")
cities = df["city"].value_counts().head(10)
print(cities)

plt.figure(figsize=(8,8))
plt.pie(cities,
        labels=cities.index,
        autopct="%1.1f%%")
plt.title("Top Host Cities")
plt.show()

print("\nMatch Result Distribution")
results = df["result"].value_counts()
print(results)

plt.figure(figsize=(8,5))
results.plot(kind="bar")
plt.title("Match Result Distribution")
plt.xlabel("Result")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

print("\nMatches by Year")

df["match_date"] = pd.to_datetime(df["match_date"])
year = df["match_date"].dt.year.value_counts().sort_index()

print(year)

plt.figure(figsize=(10,5))
plt.plot(year.index, year.values, marker="o")
plt.title("Matches Played Per Year")
plt.xlabel("Year")
plt.ylabel("Matches")
plt.grid(True)
plt.show()

print("\nTop 10 Team Appearances")

teams = pd.concat([df["team1"], df["team2"]])
appearances = teams.value_counts().head(10)

print(appearances)

plt.figure(figsize=(10,5))
appearances.plot(kind="bar")
plt.title("Top 10 Team Appearances")
plt.xlabel("Team")
plt.ylabel("Matches Played")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nBusiness Insights")
print("- Most successful teams can be identified by total wins.")
print("- Player of the Match awards highlight consistent performers.")
print("- Toss-winning patterns reveal captain strategies.")
print("- Certain venues host a significantly higher number of IPL matches.")
print("- Metro cities host the majority of IPL games.")
print("- IPL match frequency has changed across different years.")
```
